Create a while loop where the password is "python". when the password is entered the code comes out of while loop.

1 to 100 units of electicity 2 rs per unit, 100 to 200 3 rs per unit and 201 5 rs per unit

```python
units = int(input("Enter the number of units consumed: "))
totalBill = 50
if units <= 100:
    totalBill = units * 2
elif units <= 200:
    totalBill = (100 * 2) + ((units - 100) * 3)
else:
    totalBill = (100 * 2) + (100 * 3) + ((units - 200) * 5)

print(f"Total Electricity Bill: Rs {totalBill}")
```
