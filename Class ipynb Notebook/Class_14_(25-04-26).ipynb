- #### _**'~'** checks the opposite_ (like if I have to remove 2 val from a cat out of 50 we will not enter the 48 one instead we will use tilda and then put the 2 val which we do not want)

```python
df_cat = ['Frozen', 'Pantry']  # instead of putting & everytime like above this is better
df[~df['Category'].isin(df_cat)]  # isin - a membership test
```

### **sort_index()**
| Parameter | Description | Options |
|-----------|-------------|---------|
| axis | Sort by rows or columns | 0 (rows/index), 1 (columns) |
| ascending | Sort order | True (ascending - default), False (descending) |
| inplace | Modify original DataFrame | True, False (default) |
| na_position | Where to put NaN values | 'last' (default), 'first' |

```python
df.sort_index(axis = 1, ascending=False)
```

### **sort_values()**
| Parameter | Description | Options |
|-----------|-------------|---------|
| by | Column name(s) or index level(s) to sort by | string or list of strings |
| axis | Sort by rows or columns | 0 (rows/index), 1 (columns) |
| ascending | Sort order | True (ascending - default), False (descending) |
| inplace | Modify original DataFrame | True, False (default) |
| na_position | Where to put NaN values | 'last' (default), 'first' |
| ignore_index | Reset index after sorting | True, False (default) |
| key | Apply function to values before sorting | callable function |

```python
df.sort_values('Price_Dollars', ascending=False)
```
