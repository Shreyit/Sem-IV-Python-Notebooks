```python
import pandas as pd

# Step 1: Send request to Wikipedia
url = 'https://en.wikipedia.org/wiki/List_of_countries_by_life_expectancy'
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)
print(f"Status code: {response.status_code}")  # 200 = success

# Step 2: Parse the HTML
soup = BeautifulSoup(response.content, 'html.parser')

# Step 3: Find the table (Wikipedia tables often use class 'wikitable')
table = soup.find('table', class_='wikitable')

# Step 4: Extract headers (first row = headers)
headers_row = table.find_all('tr')[0]
header_names = [th.get_text(strip=True) for th in headers_row.find_all('th')]
print(f"Columns found: {header_names}")

# Step 5: Extract data rows (skip header row)
rows_data = []
for row in table.find_all('tr')[1:]:  # skip header
    cells = row.find_all(['td', 'th'])
    row_data = [cell.get_text(strip=True) for cell in cells]
    rows_data.append(row_data)

# Step 6: Create DataFrame
life_exp_df = pd.DataFrame(rows_data, columns=header_names)
life_exp_df.head(10)
```
