## **Indexing and Slicing**

#### **1-D ARRAY**

```python
import numpy as np
```

```python
product_list = ['fruit', 'vegeteable', 'cereal', 'dairy', 'eggs', 'snacks', 'beverages', 'coffee', 'tea', 'spices']
product_array = np.array(product_list)
```

```python
product_list
```

```python
product_array
```

```python
product_array.size
```

```python
len(product_array)
```

```python
product_array[-1]
```

```python
product_list[1]
```

```python
product_array[2:6]
```

#### **2-D ARRAY**

```python
product_2d_array = product_array.reshape(2,5)
```

```python
product_array
```

```python
product_2d_array[0,4]
```

```python
product_2d_array
```

```python
product_2d_array[0:, :2]
```

```python
product_2d_array[1:, 2:]
```

```python
product_2d_array[:, 1:4]
```

```python
new_2d_array = np.arange(11,23)
new_2d_array
```

```python
new_2d_array = np.arange(11,23).reshape(3,4)
new_2d_array
```

```python
new_2d_array[1:, 1:]
```

```python
new_2d_array[:2, 1:3]
```

```python
new_2d_array[:, 2:]
```

```python
new_2d_array[:, 1:2]
```
