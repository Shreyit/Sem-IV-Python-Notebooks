```python
fruit = ["apple","pineaple","orange","tomato","bluberry"]
vegetable = []
for item in fruit:
    if item == "tomato":
        continue # Skip 'tomato' - don't print it, move to next item
    else:
        print([item])
vegetable.append("tomato")
print(vegetable)
```

#### _1 to 100 units of electicity 2 rs per unit, 100 to 200 3 rs per unit and from 201 5 rs per unit_

```python
units = int(input("Enter the number of units consumed: "))
totalBill = 50
if units <= 100:
    totalBill = units * 2
elif units <= 200:
    totalBill = (100 * 2) + ((units - 100) * 3)
else:
    totalBill = (100 * 2) + (100 * 3) + ((units - 200) * 5)
    print(f"Total Electricity Bill: Rs {totalBill}")
```
