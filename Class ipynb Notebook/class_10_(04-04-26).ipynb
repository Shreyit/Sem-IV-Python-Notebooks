### **QUESTIONS**
1. Make a list comprehension where range is 0-100, element 10, convert it into array of 5 by 2. Get dim, shape, no of element, data type.
2. Make 3 arrays, in 3 different methods/fxns, 5 by 2 , range: 10-100
3. Seeding 2026, create 9 random number - 3x3 matrix,. print 1st 2 rows, then 1st column, then 2nd element of 3rd row

```python
# QUESTION 1
lst = [i for i in range(0, 100, 10)]
arr = np.array(lst).reshape(5,2)
print('array: \n', arr)
print('dimention:', arr.ndim)
print('shape: ', arr.shape)
print('size: ', arr.size)
print('data type: ', arr.dtype)
```

```python
# QUESTION 2
lst1 = [i for i in range(10, 101, 10)]
arr1 = np.array(lst1).reshape(5,2)

arr2 = np.arange(10,101,10).reshape(5,2)

lst2 = list(range(10, 101, 10))
arr3 = np.array(lst2).reshape(5,2)

arr4 = np.linspace(10,100,10).reshape(5,2)

print('array 1:\n', arr1)
print('array 2:\n', arr2)
print('array 3:\n', arr3)
print('array 4:\n', arr4)
```

```python
# QUESTION 3
x = np.random.default_rng(2026)
arr5 = x.random(9).reshape(3,3)
print('ARRAY\n',arr5)

print('\nARRAY 1st 2 rows\n',arr5[:2,])
print('\nARRAY 1st column\n',arr5[:, 0])
print('\nARRAY 2nd element of 3rd row\n',arr5[2,1])
```
