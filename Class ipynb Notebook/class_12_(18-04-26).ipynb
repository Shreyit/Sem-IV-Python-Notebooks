### **1. Type Cast (datatype)**

```python
run_time = pd.read_excel('../data/Run Times.xlsx')
run_time
```

```python
run_time.info()
```

```python
run_time.dtypes
```

```python
run_time['Fee'] = pd.to_numeric(run_time['Fee'].str.replace('$', ''))
run_time
```

```python
run_time_c = run_time.copy()
run_time_c
```

```python
run_time.dtypes
```

When converting strings to numeric, invalid data causes errors.
- `errors='raise'`: Crashes on invalid values
- `errors='coerce'`: Replaces invalid values with NaN

```python
# run_time['Warm Up Time'] = pd.to_numeric(run_time['Warm Up Time']) - this will throw error for index 2 - '3 min'
run_time['Warm Up Time'] = pd.to_numeric(run_time['Warm Up Time'], errors = 'coerce')
run_time
```

```python
run_time.dtypes
```

```python
run_time_c['Warm Up Time'] = run_time_c['Warm Up Time'].str.replace('min', '')
run_time_c
```

### **2.0 Missing Data**

```python
df_stud = pd.read_excel('../data/Student Grades.xlsx')
df_stud02 = df_stud.copy()
df_stud
```

```python
df_stud.describe()
```

```python
df_stud.isna().sum()
```

```python
df_stud[df_stud.isna().any(axis = 1)]
```

```python
df_stud
```

```python
df_stud.dropna(how = 'all', inplace = True)   # use: inplace = True to replace this in the dataframe
df_stud.dropna(subset = ['Student', 'Class'], inplace = True)
df_stud = df_stud[df_stud['Student'] != 'Bennet']
df_stud
```

```python
df_stud.isna().sum()
```

### **2.1. Filling Missing Values**

```python
df_stud['Grade'].fillna(df_stud['Grade'].mean(), inplace = True)
df_stud
```

```python
df_stud.isna().sum()
```

```python
df_stud[df_stud['Class'] == 'Freshman Seminar']
```

```python
df_stud['Year'] = np.where((df_stud['Year'].isna()) & (df_stud['Class']=='Freshman Seminar'), 'Freshman', df_stud['Year'])
df_stud
```
