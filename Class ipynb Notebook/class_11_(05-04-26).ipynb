### **Series**
A Series is a one-dimensional labeled array in pandas. It's like a column in a spreadsheet or a single column of data.

Key characteristics:
- Has an index (labels for each element)
- Can hold any data type
- Supports vectorized operations
- Similar to a dictionary or enhanced NumPy array

```python
array=np.arange(5)
pd_series= pd.Series(array, name='my first panda series')
```

```python
array
```

**Creating Series from Arrays:**
- `pd.Series(data, index, name)` creates a Series
- `data` can be a list, array, or scalar value
- `index` provides labels (defaults to 0,1,2...)
- `name` gives the Series a name for identification

```python
pd_series
```

**Creating DataFrames:**
- `pd.DataFrame(data)` creates a 2D table
- `data` can be a dictionary (keys become columns), list of lists, or other structures
- Each column is a Series
- Supports various data types per column

```python
sales= [0,23,4,3,54,23,0,2,12,54,]
sales_series=pd.Series(sales, name='sales series')
sales_series
```

```python
sales_series.astype("float")
# change the data type, but only for this cell
# pass in another variable to change type
```

```python
sales_series.astype('bool')
# zero= false, other= true
```

**Data Type Conversion:**
- `.astype(dtype)` converts Series data types
- Common types: 'int64', 'float64', 'object' (string), 'bool'
- Important for memory efficiency and operations
- Boolean conversion: 0 → False, non-zero → True

```python
sales=[0,5,155,0,518,0,1287,616,317,325]
month=['jan','feb','march','april','may','june','july','august','september','october']
sales_month= pd.Series(sales, index=month, name='Sales')
sales_month.sort_values() # sort values, sort_index() sort index 
# put (ascending=FALSE) to put in descending order, and to make it permanent, put inplace= TRUE, otherwise it sorts only for this line
```

**Sorting Series:**
- `.sort_values()` sorts by values (ascending by default)
- `.sort_index()` sorts by index labels
- `ascending=False` for descending order
- `inplace=True` modifies the original Series permanently

### **Indexing and Selection: ILOC vs LOC**
- **ILOC**: Position-based indexing (like Python lists)
  - Uses integer positions: `series.iloc[0]`, `series.iloc[0:3]`
  - Always uses numbers, regardless of index labels
- **LOC**: Label-based indexing
  - Uses index labels: `series.loc['jan']`, `series.loc['jan':'march']`
  - Works with custom index labels

```python
sales_month.iloc[0:4]
```

```python
sales_month.iloc[[2,5,9]]
# specific values
```

```python
sales_month.loc['july':'october']
```
