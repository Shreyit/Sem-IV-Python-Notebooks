1. Create and inspect an 1d array [12,24,36,48,60], tell total no of elements.

```python
import numpy as np
```

```python
# ANSWER 1
arr_1d = np.array([12, 24, 36, 48, 60])
print(arr_1d)
print('size = ',arr_1d.size)
```

```python
# ANSWER 2
arr_zeros = np.zeros((1,8), dtype=int)
print("Zeros 1d array = ", arr_zeros)
arr_ones = np.ones((1,8), dtype=int)
print("Ones 1d array = ", arr_ones)
```

```python
# ANSWER 3
arr_2d = np.zeros((4, 3), dtype=int)
print(arr_2d)
print("DIMENTION = ", arr_2d.shape)
```

```python
# ANSWER 4
lst = [45,67,89,90,55]
np_arr = np.array(lst)
np_arr
```

```python
# ANSWER 5
rng = np.random.default_rng()
rand_arr = rng.random(6).round(2)
rand_arr
```

```python
# ANSWER 6
rand = rng.integers(low = 50, high = 100, size = 10)
rand
```

```python
# ANSWER 7
rand_normal = np.random.normal(100, 10, 8)
rand_normal
```

```python
# ANSWER 8
rand_01 = np.zeros((2,3), dtype=int)
rand_02 = np.ones((3,2), dtype=int)
rand_03 = rng.integers(low=1, high=20, size=4)
print("2:3 ARRAY with 6 zeros =\n", rand_01)
print("3:2 ARRAY with 6 ones =\n", rand_02)
print("random integer ARRAY of 4 vals (range: 1-20) =\n", rand_03)
```

```python
# ANSWER 9
# numpy.random.randint(low, high=None, size=None, dtype=int)
std_id = np.random.randint(low=1000, high=9990, size=5, dtype=int)
print("Student id: ", std_id)

attendance = np.random.rand(5) * 100
print("Attendance in %: ", np.round(attendance, 2))

test_scr = np.random.normal(70, 8, 5)
print("Test Score: ", test_scr)
```
